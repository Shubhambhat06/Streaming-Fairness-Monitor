# Streaming Fairness Monitor: Window-size study

The **synthetic pilot** is measured from deterministic simulated data; the Adult experiment only appears if generated locally. None of these plots represents live Flink runtime measurements. Run all cells from the repository root or notebook directory.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
root = Path.cwd()
if not (root / "results").exists():
    root = root.parent
pilot = pd.read_csv(root / "results/study_synthetic.csv")
pilot
```

```python
fig, ax = plt.subplots(figsize=(8,4))
ax.plot(pilot.window_events, pilot.false_alarm_rate, marker="o")
ax.set(xlabel="Window size (number of events)", ylabel="False alarm rate", title="False alarms in baseline-only windows")
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8,4))
ax.plot(pilot.window_events, pilot.detection_delay_events, marker="o")
ax.set(xlabel="Window size (number of events)", ylabel="Detection delay (events after drift onset)", title="Detection delay under injected shift")
plt.show()
```

## Interpretation

A smaller sample may fluctuate more and trigger nuisance alerts. A larger sample dilutes newly injected changes and may delay detection. Here window boundaries are every 100 events, while Flink uses seconds-based processing-time windows. **The offline reference uses pre-drift windows and is different from the zero-gap live demo threshold.**
